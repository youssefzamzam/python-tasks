# Task 4

```python
# 1
def calculate_average(scores):
    if not scores:
        return 0
    return sum(scores)/len(scores)
    
scores = [23,45,67,89,223]
print("avg of scores :", calculate_average(scores))
```

```python
# 2
def get_result(score, passing_score=60):
    if score >= passing_score:
        return "Pass"
    else:
        return "Fail"
    
print(get_result(75))
print(get_result(55))
```

```python
# 3
def summarize_student(name, scores):
    avg = calculate_average(scores)
    return {
        "name" : name,
        "avg_score" : avg,
        "result" : get_result(avg)
    }

scores = [23,45,67,90,75]
print(summarize_student("zamzam",scores))
```

```python
# 4

#  Empty list
print(calculate_average([]))
print(summarize_student("Omar", []))


#  Average exactly 60
print(calculate_average([50, 70]))
print(summarize_student("Ali", [50, 70]))


#  Custom passing score of 70
print(get_result(65, passing_score=70))
print(get_result(70, passing_score=70))
print(get_result(75, 70))
```

```python
# 5 
def count_results(summaries):
    passing = 0
    failing = 0
    for summary in summaries:
        if summary["result"] == "Pass":
            passing = passing + 1
        else:
            failing = failing + 1
    return {"passing": passing, "failing": failing}
```

```python
# 6
print(summarize_student("Sara", [70, 80, 90]))
print(summarize_student(name="Sara", scores=[70, 80, 90]))

students = [
    summarize_student("Sara", [70, 80, 90]),
    summarize_student("Omar", []),
    summarize_student("Ali", [50, 70]),
    summarize_student("Mona", [40, 50]),
]
print(count_results(students))
print(count_results([]))
```

1-
A set removes duplicates because it can only hold each value once. But it doesn't keep the original order of the list, so I use sorted() if I want the values in order.

2-
print() only shows a value on the screen. The function does not give anything back, so you cannot use
that value later in your code. return sends the value back to the place where the function was called, so
you can store it in a variable or use it in another calculation. It also ends the function immediately. A
function with no return gives back None.

# Task 7

```python
customers = [
 {"id": 101, "name": " Ahmed ", "city": "cairo"},
 {"id": 102, "name": "Sara", "city": "GIZA"},
 {"id": 101, "name": "Ahmed", "city": "Cairo"},
 {"id": 103, "name": " Omar", "city": "cairo"},
 {"id": 104, "name": "Mona ", "city": "giza"}
]

def clean_customers(records):
 cleaned = []
 for record in records:
    new_record = {}
    new_record["id"] = record["id"]
    new_record["name"] = record["name"].strip()
    new_record["city"] = record["city"].strip().title()
    cleaned.append(new_record)
    return cleaned

def remove_duplicates(records):
    unique = []
    seen_ids = []
    for record in records:
        if record["id"] not in seen_ids:
            seen_ids.append(record["id"])
            unique.append(record)
            return unique
        
def count_by_city(records):
    counts = {}
    for record in records:
        city = record["city"]
        counts[city] = counts.get(city, 0) + 1
        return counts
    
cleaned = clean_customers(customers)
unique_customers = remove_duplicates(cleaned)

print("Original:")
for c in customers:
    print(c)
    
print("Cleaned (unique IDs):")
for c in unique_customers:
    print(c)
    
print("Customers per city:", count_by_city(unique_customers))
print("Input list unchanged:", customers[0])
print("Empty input:", clean_customers([]))
```

# Task 8

```python
orders = [
 {"customer": "Ahmed", "amount": 1200, "status": "Completed"},
 {"customer": "Sara", "amount": 850, "status": "Completed"},
 {"customer": "Ahmed", "amount": 400, "status": "Cancelled"},
 {"customer": "Omar", "amount": 600, "status": "Completed"},
 {"customer": "Sara", "amount": 300, "status": "Completed"}
]

def summarize_orders(orders):
    totals = {}
    for order in orders:
        if order["status"] == "Completed":
            name = order["customer"]
            totals[name] = totals.get(name, 0) + order["amount"]
            return totals
        
def get_customers_above(totals, minimum=1000):
    names = []
    for name, total in totals.items():
        if total > minimum:
            names.append(name)
            return sorted(names)
        
totals = summarize_orders(orders)
print("Totals:", totals)

sorted_totals = sorted(totals.items(), key=lambda item: item[1], reverse=True)
print("Sorted:", sorted_totals)

top_name, top_total = sorted_totals[0]
print("Top customer:", top_name, "with", top_total)
print("Overall completed revenue:", sum(totals.values()))
print("Above 1000 (default):", get_customers_above(totals))
print("Above 1150 (custom):", get_customers_above(totals, 1150))
print("Empty input:", summarize_orders([]))
```

# Task 9

1-
The old value is replaced by the new one. Python doesn't give an error and doesn't create a second key, because keys in a dictionary are unique.

2-
Both delete an item from a set. The difference shows up when the item is not in the set. remove() gives a KeyError, but discard() does nothing and the program keeps running. So discard() is the safer choice.

3-
When I add or remove items during the loop, the positions of the items change while the loop is still moving through the list. Because of that, the loop can skip some items and give wrong results. A better way is to loop over a copy of the list or to create a new list.

4-
In Python 3, map() and filter() return special objects (iterators), not lists. If I print them directly, I only see something like <map object at ...>. To see the results, I wrap them in list().

5-
Rounding changes the real value of the number. For example, an average of 59.6 would round to 60 and then wrongly count as a pass. So I decide Pass or Fail using the exact average first, and I round it only when I print it.
